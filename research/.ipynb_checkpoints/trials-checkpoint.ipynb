```python

def load_pdf_files(data):
    loader = DirectoryLoader(   ####### created instance of directoryloader
        data,
        glob="*.pdf",
        loader_cls=PyPDFLoader
    )
    documents=loader.load()
    return documents     ###list of documents
```

```python
extracted_docs = load_pdf_files(r"C:\Users\DELL\Documents\MEDIBOT\data")
```

```python
from typing import List
from langchain.schema import Document
```

```python
def filter_to_minimal_docs(docs:List[Document]) -> List[Document]:

    minimal_docs : List[Document] = [] 
    for doc in docs:
        src=doc.metadata.get("source")
        minimal_docs.append(
            Document(
               page_content=doc.page_content,
               metadata={
                "source": src
            }
        )
    )
    return minimal_docs
minimal_docs=filter_to_minimal_docs(extracted_docs)
```

```python
##################Chunking operation

def text_split(minimal_docs: List[Document]) -> List[Document]:
    text_splitter = RecursiveCharacterTextSplitter(
        chunk_size=300,
        chunk_overlap=100,
        length_function=len
    )
    texts_chunk = text_splitter.split_documents(minimal_docs)
    return texts_chunk

texts_chunk = text_split(minimal_docs)
```

```python
from langchain_community.embeddings import HuggingFaceEmbeddings 

def download_embeddings():
    model_name = "sentence-transformers/all-MiniLM-L6-v2"
    
    embeddings = HuggingFaceEmbeddings(
        model_name=model_name
        
    )                                    ###initialize the embeding model 
    return embeddings


embedding = download_embeddings()
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
import os

GEMINI_API_KEY = os.getenv("GEMINI_API_KEY")

chatModel = ChatGoogleGenerativeAI(
    model="gemini-1.5-flash",  
    google_api_key=GEMINI_API_KEY
)
```

```python
from langchain.chains import create_retrieval_chain
from langchain.chains.combine_documents import create_stuff_documents_chain
from langchain_core.prompts import ChatPromptTemplate
```

```python
from langchain.prompts import ChatPromptTemplate

system_prompt = (
    "You are a Medical assistant for question-answering tasks. "
    "Use the following pieces of retrieved context to answer the question. "
    "If you don't know the answer, say that you don't know. "
    "Use three sentences maximum and keep the answer concise.\n\n"
    "{context}"
)


prompt = ChatPromptTemplate.from_messages(
    [
        ("system", system_prompt),
        ("human", "{input}"),
    ]
)
```

```python
question_answer_chain = create_stuff_documents_chain(chatModel,prompt)
rag_chain = create_retrieval_chain(retriever,question_answer_chain)
```

```python
response = rag_chain.invoke({"input":"what is acne?"})
print(response["answer"])
```

```python
response = rag_chain.invoke({"input":"what is acromegaly?"})
print(response["answer"])
```
